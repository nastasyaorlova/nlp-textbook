# 3. Типы данных

⏩ *Если группа уверенно ответила на диагностику, объяснения этого раздела можно пролистать и перейти к заданиям в конце раздела.*

У каждого значения в Python есть **тип**. Нам понадобятся четыре основных:

| Тип | Название | Пример | Для чего |
|---|---|---|---|
| `int` | целое число | `5` | оценка в звёздах, число отзывов |
| `float` | дробное число | `3.99` | средняя оценка, процент |
| `str` | строка (текст) | `"оператор ответил быстро"` | любой текст: отзыв, пост, заголовок |
| `bool` | логическое значение | `True` / `False` | «да» или «нет»: есть ли в отзыве слово «заблокировали» |

Узнать тип можно функцией `type()`:

````python
print(type(5))
print(type(3.99))
print(type("оператор ответил быстро"))
print(type(True))
````

**Важно:** `"5"` в кавычках — это **текст**, а не число. С ним нельзя считать. Это очень пригодится, когда мы будем читать файл: всё, что читается из файла, — текст.

````python
print(5 + 4)        # числа складываются
print("5" + "4")    # строки склеиваются!
````

Чтобы превратить один тип в другой:
- `int("5")` → `5` (текст в целое число);
- `str(5)` → `"5"` (число в текст).

````python
grade = "5"                # так оценка выглядит, когда мы читаем её из файла
print(int(grade) + 1)      # превратили в число — можно считать
````

Удобнее всего вставлять значения в текст с помощью **f-строки**: перед кавычками ставится буква `f`, а переменные пишутся в фигурных скобках `{}`.

````python
city = "Москва"
grade = 5

print(f"Клиент из города {city} поставил оценку {grade}")
````

Строка — главный тип данных в нашем курсе: каждый отзыв, пост и новость — это строка.

## 4.1. Длина строки
`len()` считает количество символов, включая пробелы и знаки препинания.

````python
review = "Оператор ответил быстро, но кэшбэк так и не начислили!"
print(len(review))
````

## 4.2. Индексы: доступ к отдельным символам

Каждый символ в строке имеет номер — **индекс**. **Нумерация начинается с нуля!**

```
 Б  а  н  к
 0  1  2  3      ← индексы с начала
-4 -3 -2 -1      ← индексы с конца
```

````python
word = "Банк"
print(word[0])     # первый символ
print(word[1])     # второй символ
print(word[-1])    # последний символ
````

## 4.3. Срезы: кусок строки

`строка[начало:конец]` — символы с индекса `начало` до индекса `конец`, **не включая** конец. Срезы удобны, чтобы показать начало длинного отзыва.

````python
date = "15.01.2026"
print(date[0:2])    # день: символы 0 и 1
print(date[3:5])    # месяц
print(date[6:])     # год: с 6 до конца
````

## 4.4. Методы строк

**Метод** — действие, которое умеет делать строка. Пишется через точку: `строка.метод()`.

| Метод | Что делает | Пример | Результат |
|---|---|---|---|
| `.lower()` | всё в нижний регистр | `"СБП".lower()` | `"сбп"` |
| `.upper()` | всё в верхний регистр | `"сбп".upper()` | `"СБП"` |
| `.replace(a, b)` | заменяет `a` на `b` | `"кешбэк".replace("е", "э")` | `"кэшбэк"` |
| `.strip()` | убирает пробелы и переносы по краям | `"  карта  ".strip()` | `"карта"` |
| `.count(a)` | сколько раз встречается `a` | `"ааб".count("а")` | `2` |
| `.startswith(a)` | начинается ли с `a` | `"Банк".startswith("Б")` | `True` |
| `.split()` | делит на слова → список | `"карту заблокировали".split()` | `["карту", "заблокировали"]` |

Посмотрим на практике:

````python
review = "  Карту ЗАБЛОКИРОВАЛИ без объяснений!  "

print(review.lower())
print(review.strip())
print(review.replace("ЗАБЛОКИРОВАЛИ", "заблокировали"))
print(review.count("о"))
````

**Важно:** методы не меняют исходную строку, а создают новую. Чтобы сохранить результат, его нужно положить в переменную:

````python
review = "Карту ЗАБЛОКИРОВАЛИ"
review.lower()          # результат никуда не сохранили
print(review)           # строка не изменилась

review = review.lower() # сохранили результат в ту же переменную
print(review)
````

Методы можно применять один за другим — **цепочкой**:

````python
review = "  Карту ЗАБЛОКИРОВАЛИ!  "
clean = review.strip().lower()
print(clean)
````

## 4.5. Есть ли слово в тексте?

Оператор `in` проверяет, содержится ли одна строка в другой. Результат — `True` или `False`.

````python
review = "оператор ответил быстро, но кэшбэк так и не начислили"

print("кэшбэк" in review)
print("кредит" in review)
````

`in` различает большие и маленькие буквы. Поэтому текст перед поиском обычно приводят к нижнему регистру:

````python
review = "СБП не работает уже третий день"

print("сбп" in review)            # False: в тексте «СБП» большими буквами
print("сбп" in review.lower())    # True
````

## 5.2. От текста к списку и обратно

Это главная связка в анализе текстов:
- `строка.split()` — делит текст на слова по пробелам и возвращает **список**;
- `строка.split(";")` — делит по любому другому разделителю, например по `;`;
- `" ".join(список)` — склеивает список обратно в **строку**.

````python
review = "карту заблокировали без объяснения причин"

words = review.split()
print(words)
print("Слов в отзыве:", len(words))

print(" ".join(words))
print(" | ".join(words))    # склеить можно любым разделителем
````

**Это очень важно для датасета.** Каждая строка нашего файла с отзывами выглядит так: дата, город, оценка и текст, разделённые точкой с запятой. Чтобы достать отдельные части, строку делят по `;`:

````python
line = "15.01.2026;Москва;5;оператор ответил быстро"

parts = line.split(";")
print(parts)
print("Дата:", parts[0])
print("Оценка:", parts[2])
````

## 5.3. Есть ли элемент в списке?

`in` работает и со списками. Но здесь он ищет **элемент целиком**:

````python
words = ["карту", "заблокировали", "без", "объяснения"]

print("карту" in words)    # True: такой элемент есть
print("карт" in words)     # False: элемента «карт» в списке нет
````

# 6. Условия: `if` / `elif` / `else`

Условия позволяют программе принимать решения: «**если** в отзыве есть слово "заблокировали", **то** отметить его как жалобу на блокировку».

```python
if условие:
    что делать, если условие верно
else:
    что делать, если неверно
```

**Обратите внимание:**
- после условия ставится двоеточие `:`;
- строки внутри условия сдвигаются на **4 пробела** (Colab делает это сам после двоеточия). Отступ показывает, какие строки относятся к условию.

````python
review = "карту заблокировали без объяснения причин"

if "заблокир" in review:
    print("Жалоба на блокировку")
else:
    print("О блокировке не пишут")
````

## 6.1. Сравнения

| Знак | Значение | Пример |
|---|---|---|
| `==` | равно | `grade == 5` |
| `!=` | не равно | `grade != 5` |
| `>` / `<` | больше / меньше | `grade > 3` |
| `>=` / `<=` | больше или равно / меньше или равно | `grade >= 4` |

⚠️ `=` — это «положить в переменную», а `==` — «проверить на равенство».

Если вариантов больше двух, используется `elif` («иначе, если»):

````python
grade = 3

if grade >= 4:
    print("Положительный отзыв")
elif grade == 3:
    print("Нейтральный отзыв")
else:
    print("Отрицательный отзыв")
````

# 7. Цикл `for`: повторяем действие для каждого элемента

Отзывов у нас 500. Писать код для каждого отдельно невозможно. Цикл `for` берёт элементы списка **по очереди** и для каждого выполняет одни и те же действия.

```python
for элемент in список:
    что сделать с элементом
```

````python
reviews = ["оператор ответил быстро", "карту заблокировали", "удобное приложение"]

for review in reviews:
    print(review)
````

Как это работает:
1. `review` = `"оператор ответил быстро"` → выполняется `print`
2. `review` = `"карту заблокировали"` → выполняется `print`
3. `review` = `"удобное приложение"` → выполняется `print`
4. Элементы закончились — цикл завершён.

Имя переменной (`review`) вы выбираете сами. Обычно берут слово в единственном числе: `for review in reviews`, `for word in words`, `for grade in grades`.

Внутри цикла можно делать что угодно — например, считать длину каждого отзыва:

````python
for review in reviews:
    print(review, "—", len(review.split()), "слова")
````

# 8. Словари: считаем слова

**Словарь** хранит пары «**ключ → значение**», как настоящий словарь: слово → перевод. Пишется в фигурных скобках.

````python
grades_by_city = {"Москва": 4.1, "Пермь": 3.2}

print(grades_by_city["Москва"])     # значение по ключу
grades_by_city["Казань"] = 4.5      # добавить новую пару
print(grades_by_city)
````

Для анализа текстов словарь особенно полезен: в нём удобно хранить **частоты слов** — слово → сколько раз оно встретилось.

Алгоритм подсчёта:
1. Заводим пустой словарь.
2. Идём по словам циклом.
3. Если слово уже есть в словаре, увеличиваем его счётчик на 1. Если нет — добавляем со значением 1.

````python
text = "деньги списали деньги не вернули карту заблокировали деньги"
words = text.split()

freq = {}
for word in words:
    if word in freq:
        freq[word] = freq[word] + 1
    else:
        freq[word] = 1

print(freq)
````

Задача подсчёта частот встречается так часто, что для неё есть готовый инструмент — `Counter`. Он делает то же самое в одну строку:

````python
from collections import Counter

text = "деньги списали деньги не вернули карту заблокировали деньги"
freq = Counter(text.split())

print(freq)
print(freq.most_common(2))    # два самых частых слова
````

Посмотрим, что получится на тексте со знаками препинания:

````python
text = "оператор ответил быстро. оператор вежливый, вопрос решили быстро. спасибо оператор"

freq = Counter(text.split())
print(freq.most_common(3))
````

💡 Посмотрите на результат внимательно: `"быстро."` и `"быстро"` посчитались как **разные** слова из-за точки. Знаки препинания нужно убирать перед подсчётом. Как это сделать, покажем в разделе 11, а на шестом семинаре научимся делать правильно.

# 9. Функции: свои команды

Мы уже пользовались готовыми функциями: `print()`, `len()`, `sum()`. Можно написать и свою — чтобы не повторять один и тот же код много раз.

```python
def имя_функции(аргумент):
    действия
    return результат
```

- `def` — «определить функцию»;
- в скобках — **аргумент**: что функция получает на вход;
- `return` — что функция возвращает в результате.

````python
def remove_punctuation(text):
    for mark in ".,!?:;()":
        text = text.replace(mark, " ")
    return text

print(remove_punctuation("оператор ответил быстро. спасибо!"))
````

Функция сама ничего не делает, пока её не **вызвали**: `remove_punctuation(...)`. Теперь убрать знаки препинания можно из любого отзыва одной строкой.

У функции может быть несколько аргументов:

````python
def percent(part, total):
    return part / total * 100

print(percent(107, 500))
print(percent(353, 500))
````

## 10.4. Делим строки на части — и находим ловушку

Разделим каждую строку по `;` и проверим, что у всех получилось ровно 4 части: дата, город, оценка, отзыв.

````python
broken = 0
for line in lines:
    parts = line.split(";")
    if len(parts) != 4:
        broken = broken + 1

print("Строк, которые разделились неправильно:", broken)
````

Шесть строк разделились неправильно. Посмотрим на одну из них:

````python
for line in lines:
    parts = line.split(";")
    if len(parts) != 4:
        print("Частей:", len(parts))
        print(line[:300])
        break      # break останавливает цикл: нам достаточно одного примера
````

**Причина:** клиент сам поставил в тексте отзыва точку с запятой, и `split` разрезал отзыв на куски.

**Решение:** отзыв стоит в строке последним. Поэтому можно попросить `split` разделить строку только по **первым трём** `;`, а всё остальное оставить одним куском: `line.split(";", 3)`.

````python
broken = 0
for line in lines:
    parts = line.strip().split(";", 3)
    if len(parts) != 4:
        broken = broken + 1

print("Сломанных строк:", broken)
````

Такие ловушки встречаются в реальных данных постоянно. Поэтому после загрузки всегда проверяйте, что данные прочитались правильно.

````python
by_year = {}
for i in range(len(dates)):
    year = dates[i][6:]
    if year not in by_year:
        by_year[year] = []          # первый отзыв этого года: создаём пустой список
    by_year[year].append(grades[i])

for year in sorted(by_year):
    year_grades = by_year[year]
    print(year, "— отзывов:", len(year_grades), "| средняя оценка:", round(sum(year_grades) / len(year_grades), 2))
````

## 11.3. Кто пишет длиннее

````python
bad_lengths = []
good_lengths = []

for i in range(len(texts)):
    words_count = len(texts[i].split())
    if grades[i] <= 2:
        bad_lengths.append(words_count)
    elif grades[i] >= 4:
        good_lengths.append(words_count)

print("Отрицательные отзывы:", round(sum(bad_lengths) / len(bad_lengths)), "слов в среднем")
print("Положительные отзывы:", round(sum(good_lengths) / len(good_lengths)), "слов в среднем")
````

💬 Недовольные клиенты пишут примерно в полтора раза длиннее: они подробно рассказывают историю проблемы. Для PR-службы это значит, что в отрицательных отзывах больше конкретики, с которой можно работать.

## 11.4. О чём пишут: темы отзывов

Составим словарь тем: для каждой темы — список частей слов, по которым её можно найти. Части слов, а не целые слова, нужны, чтобы находить разные формы: «блокир» найдёт и «заблокировали», и «блокировка».

Затем для каждой темы посчитаем, сколько отзывов её упоминают и какую среднюю оценку ставят авторы этих отзывов.

````python
themes = {
    "блокировки":  ["блокир"],
    "переводы":    ["перевод", "сбп"],
    "кредиты":     ["кредит", "рассрочк", "займ"],
    "приложение":  ["приложени"],
    "кэшбэк":      ["кэшб", "кешб"],
    "вклады":      ["вклад", "накопит"],
}

def mentions(text, keywords):
    "Возвращает True, если в тексте есть хотя бы одно из ключевых слов"
    for keyword in keywords:
        if keyword in text:
            return True
    return False

for theme in themes:
    theme_grades = []
    for i in range(len(texts)):
        if mentions(texts[i], themes[theme]):
            theme_grades.append(grades[i])
    average = round(sum(theme_grades) / len(theme_grades), 2)
    print(f"{theme:12} упоминаний: {len(theme_grades):3}   средняя оценка: {average}")
````

💬 **Что мы видим:** темы чётко делятся на «больные» и «сильные».
- **Больные места:** блокировки (средняя оценка около 2) и переводы (около 2,7). Об этом клиенты пишут, когда у них что-то сломалось.
- **Сильные стороны:** кэшбэк и вклады (около 4,7–4,8). На них банк может строить коммуникацию.

Это уже готовый вывод для PR-отчёта. Обратите внимание: результат зависит от того, какие слова мы включили в словарь тем. Составляет его исследователь, поэтому словарь нужно проверять, читая отзывы.

## 11.6. Какими словами пишут довольные и недовольные клиенты

Уберём знаки препинания, числа и служебные слова, а потом посчитаем самые частые слова отдельно для отрицательных и положительных отзывов.

````python
# Служебные слова, которые есть в любом тексте и ничего не говорят о его содержании
stopwords = ["и", "в", "во", "не", "что", "на", "я", "с", "со", "как", "а", "то", "все", "она",
             "так", "его", "но", "да", "ты", "к", "у", "же", "вы", "за", "бы", "по", "только",
             "мне", "было", "вот", "от", "меня", "еще", "ещё", "нет", "о", "из", "ему", "когда",
             "даже", "ну", "ли", "если", "уже", "или", "ни", "быть", "был", "до", "там", "для",
             "мы", "их", "чем", "была", "без", "раз", "тоже", "себе", "под", "будет", "это",
             "этот", "мой", "моей", "при", "очень", "через", "они", "после", "который", "которые"]

def clean(text):
    "Убирает знаки препинания, числа и служебные слова. Возвращает список слов."
    words = remove_punctuation(text).split()
    result = []
    for word in words:
        if word.isalpha() and word not in stopwords:    # isalpha(): слово состоит только из букв
            result.append(word)
    return result

print(clean("оператор ответил быстро, вопрос решили за 5 минут. спасибо!"))
````

````python
bad_words = []
good_words = []

for i in range(len(texts)):
    if grades[i] <= 2:
        bad_words = bad_words + clean(texts[i])
    elif grades[i] >= 4:
        good_words = good_words + clean(texts[i])

print("ОТРИЦАТЕЛЬНЫЕ ОТЗЫВЫ:")
print(Counter(bad_words).most_common(15))
print()
print("ПОЛОЖИТЕЛЬНЫЕ ОТЗЫВЫ:")
print(Counter(good_words).most_common(15))
````

💬 **Что мы видим:**
- В положительных отзывах — «оператор», «быстро», «спасибо», «поддержки»: клиенты благодарят за работу чата.
- В отрицательных — «деньги», «счет», «средств», «документы», «дней»: клиенты пишут о том, что их деньги недоступны и проблема решается долго.
- Слова «банк» и «банка», «чат» и «чате» посчитались как **разные**. Компьютер не знает, что это одно слово в разных формах. Эту проблему решает **лемматизация** — приведение слов к начальной форме. Ей посвящён шестой семинар.
